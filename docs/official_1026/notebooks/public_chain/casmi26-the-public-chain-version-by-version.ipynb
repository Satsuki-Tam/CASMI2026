in two panels, split at 8 October, so that the text stays readable. Only the main line and its branches are drawn.
The family tree printed under the graphs has every notebook.

```python
from matplotlib.patches import FancyBboxPatch

show = nbs[nbs.show].copy()
firstdate = ver.groupby("ref").date.min()
bestv = ver.groupby("ref").score.max()
setter_refs = set(setters.ref)
copies = nbs[nbs.how == "copy"].parent.value_counts().to_dict()

def draw_lineage(keys, t0, t1, W, title):
    """Boxes start at a notebook's first scored version; a parent drawn in the other panel is a grey stub."""
    keys = [k for k in keys]
    ghosts = sorted({show.loc[k, "parent"] for k in keys
                     if isinstance(show.loc[k, "parent"], str) and show.loc[k, "parent"] in show.index
                     and show.loc[k, "parent"] not in keys}, key=lambda g: firstdate[g])
    xs0 = mdates.date2num(t0)
    lanes, pos = {}, {}
    def free(l, a, b):
        return all(b <= c or a >= d for c, d in lanes.get(l, []))
    for g in ghosts:                                   # stubs in the left margin
        lane = next(l for l in range(200) if free(l, xs0 - W, xs0))
        lanes.setdefault(lane, []).append((xs0 - W, xs0)); pos[g] = (xs0 - W, lane)
    kids = {k: [] for k in keys}
    roots = []
    for k in keys:
        p = show.loc[k, "parent"]
        (kids[p] if isinstance(p, str) and p in kids else roots).append(k)
    order = []
    def walk(k):
        order.append(k)
        for c in sorted(kids[k], key=lambda c: firstdate[c]):
            walk(c)
    for r in sorted(roots, key=lambda c: firstdate[c]):
        walk(r)
    for k in order:
        p = show.loc[k, "parent"]
        x0 = mdates.date2num(firstdate[k])
        has_p = isinstance(p, str) and p in pos
        lane, span = None, (x0, x0 + W)
        if has_p:
            pl = pos[p][1]
            if x0 >= pos[p][0] + W - 0.02 and free(pl, pos[p][0] + W, x0 + W):
                lane, span = pl, (pos[p][0] + W, x0 + W)   # same lane: keep the arrow's path clear
            else:
                for d in range(1, 80):
                    for l in (pl + d, pl - d):
                        if l >= 0 and lane is None and free(l, x0, x0 + W):
                            lane = l
                    if lane is not None:
                        break
        else:
            lane = next(l for l in range(200) if free(l, x0, x0 + W))
        lanes.setdefault(lane, []).append(span)
        pos[k] = (x0, lane)
    n_lanes = max(l for _, l in pos.values()) + 1
    fig, ax = plt.subplots(figsize=(16, 0.78 * n_lanes + 1.3))
    ax.grid(False)
    for k, (x0, lane) in pos.items():
        y, ghost = -lane, k in ghosts
        hot = (k in setter_refs) and not ghost
        ax.add_patch(FancyBboxPatch((x0, y - 0.36), W * 0.93, 0.72, boxstyle="round,pad=0.02,rounding_size=0.1",
                                    fc=SURF if ghost else "#ffffff", ec=BLUE if hot else "#a9a89f",
                                    lw=2.0 if hot else 1.0, ls="--" if ghost else "-", zorder=3))
        nm = safe(nbs.loc[k, "author"]) or k.split("/")[0]
        extra = f"  +{copies[k]} copies" if copies.get(k) and not ghost else ""
        c1 = "#9a998f" if ghost else INK2
        ax.text(x0 + W * 0.04, y + 0.17, nm[:24], fontsize=8.6, color=c1, va="center", zorder=4)
        ax.text(x0 + W * 0.04, y - 0.01, nbs.loc[k, "short"][:23], fontsize=9.0, color=c1 if ghost else INK,
                va="center", zorder=4, weight="bold")
        ax.text(x0 + W * 0.04, y - 0.20, ("from the panel above" if ghost else
                                         f"{bestv[k]:.3f} · {firstdate[k]:%d %b}{extra}"),
                fontsize=8.3, color=c1, va="center", zorder=4, style="italic" if ghost else "normal")
    for k, (x0, lane) in pos.items():
        if k in ghosts:
            continue
        for p, style in ((show.loc[k, "parent"], "-"), (show.loc[k, "parent2"], "--")):
            if isinstance(p, str) and p in pos:
                px, pl = pos[p]
                if px + W * 0.93 <= x0 + 0.01:
                    a, b = (px + W * 0.93, -pl), (x0, -lane)
                else:
                    a, b = (px + W * 0.45, -pl - 0.36 if pl < lane else -pl + 0.36), \
                           (x0 + W * 0.2, -lane + 0.36 if pl < lane else -lane - 0.36)
                ax.annotate("", xy=b, xytext=a, zorder=2,
                            arrowprops=dict(arrowstyle="-|>", color="#8f8e86", lw=1.1, ls=style, shrinkA=1, shrinkB=1,
                                            alpha=0.6 if style == "--" else 1.0,
                                            connectionstyle="arc3,rad=0" if pl == lane else "arc3,rad=-0.1"))
    ax.set_xlim(xs0 - (W + 0.2 if ghosts else 0.3), mdates.date2num(t1) + 0.2)
    ax.set_ylim(-n_lanes + 0.5, 0.65)
    ax.xaxis_date()
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))
    ax.set_yticks([]); ax.spines["left"].set_visible(False)
    ax.set_title(title, loc="left", fontsize=12.5)
    fig.tight_layout()
    plt.show()

split = pd.Timestamp("2026-10-08")
early = [k for k in show.index if firstdate[k] < split]
late_nb = [k for k in show.index if firstdate[k] >= split]
draw_lineage(early, firstdate[early].min(), split + pd.Timedelta(days=1.5), 3.3,
             "Who built on whom, up to 8 October (each box starts at the notebook's first scored version)")
draw_lineage(late_nb, split, firstdate[late_nb].max() + pd.Timedelta(days=1.75), 1.6,
             "Who built on whom, from 8 October (dashed grey boxes are parents drawn in the panel above)")

def tree(k, depth):
    r = nbs.loc[k]
    print(f"{'   ' * depth}{'└─ ' if depth else ''}{r.author} · {r.short}  best {bestv[k]:.3f}  [{r.how}]")
    for c in sorted([c for c in nbs.index if nbs.loc[c, "parent"] == k], key=lambda c: firstdate[c]):
        tree(c, depth + 1)
print("Every notebook in the table, as a family tree (indent = built on the line above):")
for k in sorted([k for k in nbs.index if not isinstance(nbs.loc[k, "parent"], str)], key=lambda c: firstdate[c]):
    tree(k, 0)
H.update(n_copies=str(sum(copies.values())), n_copy_targets=str(len(copies)),
         n_roots=str(sum(1 for k in nbs.index if not isinstance(nbs.loc[k, "parent"], str))))
print(f"\n{H['n_roots']} roots; {H['n_copies']} notebooks start as exact code copies of {H['n_copy_targets']} others")
```

## 4. What a step is worth

The leaderboard page says the public score uses about 33% of the test set. The data page says the test set
has about 400 molecules. So a public score is the mean of about 132 reciprocal ranks, which gives three
yardsticks:

- **One molecule.** Finding one more molecule at rank 1 instead of missing it moves a public score by 0.0076.
  Moving one from rank 2 to rank 1 moves it by 0.0038. The board shows three decimals, so a displayed step of
  0.001 can be anything from almost nothing to 0.002.
- **One standard error.** A score near 0.465 on 132 molecules has a standard error of at most 0.043
  (the largest a mean of numbers between 0 and 1 can have), and 0.036 to 0.042 under the hit-shape model from
  my earlier CASMI noise notebook.
- **A re-run.** 33 groups of versions in the table have identical code, identical mounted dataset versions
  and the same image. 30 of them land on exactly the same displayed score. The other 3
  differ by 0.002 and 0.004 and 0.006.

```python
n_pub = round(TEST_MOLECULES * PUBLIC_FRACTION)
one_mol, r21 = 1 / n_pub, 0.5 / n_pub
m = frontier
se_bound = math.sqrt(m * (1 - m) / n_pub)

def se_model(m, s, q, n=n_pub, K=25):
    """Hits land at rank 1 with share s; the rest spread over ranks 2..K with geometric weight q."""
    w = np.array([q ** (k - 2) for k in range(2, K + 1)]); w /= w.sum()
    rr = 1 / np.arange(2, K + 1)
    e1 = s + (1 - s) * (w * rr).sum()
    e2 = s + (1 - s) * (w * rr ** 2).sum()
    p = m / e1
    return math.sqrt(max(p * e2 - m * m, 0) / n) if p <= 1 else float("nan")
grid = [g for g in (se_model(m, s, q) for s in (0.5, 0.6, 0.7, 0.8) for q in (0.5, 0.7, 0.9)) if g == g]

steps = ver.dropna(subset=["step"]).copy()
steps["abs"] = steps.step.abs()
spreads = []
for g in RERUN_GROUPS:
    sc = [score_of[(r, int(v))] for r, v in g]
    spreads.append((max(sc) - min(sc), g))
rerun_max = max(s for s, _ in spreads) if spreads else float("nan")
H.update(n_test=str(TEST_MOLECULES), pub_pct=str(round(PUBLIC_FRACTION * 100)), n_pub=str(n_pub),
         one_mol=f"{one_mol:.4f}", r21=f"{r21:.4f}", tick="0.001", tick2="0.002", se_bound=f"{se_bound:.3f}",
         se_lo=f"{min(grid):.3f}", se_hi=f"{max(grid):.3f}", n_steps=str(len(steps)),
         med_step=f"{steps['abs'].median():.3f}",
         pct_lt_mol=str(round(100 * (steps['abs'] < one_mol).mean())),
         pct_le_002=str(round(100 * (steps['abs'] <= 0.002 + 1e-9).mean())),
         pct_lt_se=str(round(100 * (steps['abs'] < se_bound).mean())),
         n_groups=str(len(spreads)), n_rerun_moved=str(sum(s > 1e-9 for s, _ in spreads)),
         n_rerun_same=str(sum(s <= 1e-9 for s, _ in spreads)),
         rerun_gaps=" and ".join(f"{s:.3f}" for s, _ in sorted(spreads, key=lambda t: t[0]) if s > 1e-9),
         rerun_max=f"{rerun_max:.3f}", late_units=f"{late.gain.sum() * n_pub:.1f}")
print(f"public molecules ~ {TEST_MOLECULES} x {PUBLIC_FRACTION} = {n_pub}")
print(f"one molecule, missed -> rank 1: {H['one_mol']};  rank 2 -> rank 1: {H['r21']}")
print(f"a displayed step of {H['tick']} is a true step between 0.000 and {H['tick2']}")
print(f"standard error of a public score near {frontier:.3f}: at most {H['se_bound']}; model range {H['se_lo']} to {H['se_hi']}")
print(f"\nsteps between a version and its parent: {H['n_steps']}; median size {H['med_step']}")
print(f"  smaller than one molecule ({H['one_mol']}): {H['pct_lt_mol']}%")
print(f"  no bigger than {H['tick2']}: {H['pct_le_002']}%")
print(f"  smaller than one standard error ({H['se_bound']}): {H['pct_lt_se']}%")
print(f"  frontier gains since {H['late_start']}: {H['late_total']} in total = {H['late_units']} reciprocal-rank units")
print(f"\nre-run groups (same code, datasets, image): {H['n_groups']}; same displayed score: {H['n_rerun_same']}; "
      f"different: {H['n_rerun_moved']} (gaps {H['rerun_gaps']}); largest {H['rerun_max']}")
for s, g in sorted(spreads, key=lambda t: -t[0]):
    xs = sorted(g, key=lambda x: score_of[(x[0], int(x[1]))])
    print(f"  spread {s:.3f}: " + "; ".join(f"{who(r)} v{v} {score_of[(r, int(v))]:.3f}" for r, v in xs[:3])
          + (f" (+{len(xs) - 3} more)" if len(xs) > 3 else ""))

print("\nThe same one change, on different parents:")
same_rows = []
for change, members in SAME_CHANGE.items():
    print(f"  {change}")
    for r, v in members:
        x = ver[(ver.ref == r) & (ver.v == v)].iloc[0]
        same_rows.append((change, x.step))
        print(f"     {plabel(x):48s} -> {who(r)} v{v}:  {x.pscore:.3f} -> {x.score:.3f}  ({x.step:+.3f})")
rel = [s for c, s in same_rows if c == "PubChem promotion threshold rel > 250 → 150"]
H.update(rel_n=str(len(rel)), rel_lo=f"{min(rel):+.3f}", rel_hi=f"{max(rel):+.3f}")
print(f"\n  rel > 250 -> 150 on {H['rel_n']} parents: steps from {H['rel_lo']} to {H['rel_hi']}")
for key, name in (("n1c", "N1 challenger: proxies kept where ICEBERG exists, joint re-order"),
                  ("pcl", "PubChemLite CC0 candidates in engine 2")):
    xs = sorted(s for c, s in same_rows if c == name)
    H[key + "_n"] = str(len(xs))
    H[key + "_steps"] = " and ".join(f"{s:+.3f}" for s in xs) if xs else "none"
    print(f"  {name} on {H[key + '_n']} parents: steps {H[key + '_steps']}")

fig, ax = plt.subplots(figsize=(11.5, 4.2))
u = (steps["abs"] * n_pub).clip(upper=10.25)
ax.hist(u, bins=np.arange(0, 10.6, 0.25), color=GREY, edgecolor=SURF, linewidth=1.5, zorder=2)
ymax = ax.get_ylim()[1]
for x, txt, col, ls, yf, ha in ((0.5, "rank 2 → 1", INK2, ":", 0.52, "right"),
                                (1.0, "one molecule:\nmissed → rank 1", INK2, ":", 0.62, "left"),
                                (rerun_max * n_pub, "largest\nre-run gap", BLUE, "--", 0.82, "left"),
                                (se_bound * n_pub, "one standard\nerror (bound)", INK, "-", 0.97, "left")):
    if x == x:
        ax.axvline(x, color=col, ls=ls, lw=1.4, zorder=3)
        ax.text(x + (0.08 if ha == "left" else -0.08), ymax * yf, txt, color=col, fontsize=8.5, va="top", ha=ha)
ax.set_xlabel(f"size of the step in reciprocal-rank units on {n_pub} public molecules (the last bar holds 10 and more)")
ax.set_ylabel("steps")
ax.set_title("How big the steps between a version and its parent are", loc="left", fontsize=12)
fig.tight_layout()
plt.show()
```

**What I take from this.** Of the 470 steps in the table, 69% are smaller than one molecule found at
rank 1, 46% are no bigger than 0.002, and 95% are smaller than one standard error. Since
9 October the frontier rose 17 times, and all of those gains together come to 4.2 reciprocal-rank
units. Identical code with identical inputs usually lands on the same displayed score, but not always: the
largest gap between two such runs here is 0.006.

So when I chase a public increment of one or two thousandths, I am chasing less than one molecule moving a rank or
two, on a board where identical code can land that far apart. The board shows the step. It cannot tell me what the
change does on molecules it does not score. The one-line promotion change above, rel > 250 → 150, moved 4
different parents by +0.003 to +0.006. That is the same change doing the same thing to the same public
molecules on different bodies. It is consistent, but it is not a second sample.

Two later changes in the table were each made on two separate lineages, and they show both cases. The N1
challenger moved its 2 parents by +0.008 and +0.008, about one molecule each time: consistent, and still read
on the same public molecules. PubChemLite in engine 2 moved its 2 parents by -0.003 and +0.004: one change, two
signs. That is what an effect smaller than one molecule looks like on this board.
